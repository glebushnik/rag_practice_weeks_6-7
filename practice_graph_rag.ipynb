# GraphRAG для поддержки SaaS-инцидентов

Я подготовил этот ноутбук как маршрут работы. Данные лежат в `data/knowledge_base/`, вопросы — в `evaluation/questions.jsonl`. Здесь нет готовых фрагментов решения: заполняй пустые ячейки своим кодом и коротко объясняй выборы. Начни с [README](README.md).

## 1. Среда и данные

Подготовь Colab, зависимости и доступ к файлам репозитория. Покажи, какие форматы обнаружены и сколько источников доступно.

## 2. Чтение и проверка источников

Извлеки содержание из Markdown, DOCX, PDF и CSV. Сохрани ID документа и достаточно метаданных для проверки цитат. Покажи несколько результатов чтения и отметь проблемы, если они возникли.

## 3. Подготовка к поиску

Подготовь корпус для retrieval. Зафиксируй свой выбор единицы поиска и объясни, как после поиска вернуться к исходному документу.

## 4. Локальная модель

Запусти выбранную небольшую LLM внутри Colab и проверь один короткий запрос. Запиши название модели и ограничения среды. Внешний платный API не используется.

## 5. Базовый RAG

Собери первую версию ответа по найденным фрагментам. Для каждого вопроса сохрани текст ответа, найденные ID источников и время выполнения. Проверь несколько вопросов из `evaluation/questions.jsonl`.

## 6. Граф знаний

Выдели сущности и отношения, которые нужны для этого кейса. Построй граф по корпусу. Покажи его размер и несколько связей вместе с ID документов, из которых они получены. Объясни, как ты проверял спорные связи.

## 7. Поиск с учётом связей

Добавь использование графа при сборе доказательств для ответа. На одинаковых вопросах покажи, какие источники нашлись в базовой и графовой версиях.

## 8. Процесс в LangGraph

Оформи многошаговую обработку вопроса через состояние, узлы и переходы. В процессе должен быть явный предел шагов и путь для ответа при недостатке данных. Покажи трассу хотя бы для одного вопроса.

## 9. Проверка качества

Составь небольшой собственный эталонный набор и оцени обе версии одинаково. Посчитай показатели из README: Evidence Recall@5, покрытие фактов, точность ссылок, корректный отказ, проверку связей, время и число шагов. Сохрани ответы на все вопросы из `evaluation/questions.jsonl`.

## 10. Разбор результатов

Где граф изменил результат? Где он добавил лишние шаги или шум? Разбери не меньше двух конкретных вопросов: отдельно качество найденных документов и отдельно качество финального ответа.

## 11. Вывод

Напиши короткий вывод: в каких вопросах GraphRAG был полезен, какие ограничения связаны с корпусом и маленькой LLM, и что бы ты изменил при появлении новых внутренних документов.
